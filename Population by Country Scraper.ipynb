## Scraping Data from a Real Website

```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://www.worldometers.info/world-population/population-by-country/'
page = requests.get(url)

page.encoding = "utf-8"

soup = BeautifulSoup(page.text, 'html')
```

```python
print(soup)
```

```python
soup.find('table')
```

```python
# similar output
soup.find_all('table', class_= 'datatable w-full border border-zinc-200' )
```

```python
table = soup.find('table')
```

```python
world_titles = soup.find_all('th')
```

```python
world_titles
```

```python
world_table_titles = [title.text for title in world_titles]

print(world_table_titles)
```

```python
import pandas as pd

df = pd.DataFrame(columns = world_table_titles)

df
```

```python
column_data = soup.find_all('tr')
```

```python
for row in column_data[1:]:
    row_data = row.find_all('td')
    individual_row_data = [data.text for data in row_data]
    print(individual_row_data)
```

```python
for row in column_data[1:]:
    row_data = row.find_all('td')
    individual_row_data = [data.text.strip().replace("−", "-") for data in row_data]

    length = len(df)
    df.loc[length] = individual_row_data
```
